```python
for f in file_list:
    fsplit = f.split("_")
    if len(fsplit)==5:
        tmpdf =  pd.read_csv("./voxels/"+f, index_col=0)
        covmat = tmpdf.T.cov()
        if covmat.shape[0]<10:
            continue
        
        print(f)    
        z = linkage(covmat, 'complete')
        cov=sns.clustermap(covmat, row_linkage=z, col_linkage=z, 
                       cmap=sns.color_palette("flare",as_cmap=True))
        cov.savefig("./figs/covariances/"+"_".join([fsplit[0],fsplit[1],fsplit[4][:-4]])+"_clustering_cov.jpg")
        
        fig = plt.figure()
        dend = dendrogram(z)
        fig.savefig("./figs/dendrograms/"+"_".join([fsplit[0],fsplit[1],fsplit[4][:-4]])+"_clustering_cov_dendrogram.jpg")
        
        np.save("./"+"_".join([fsplit[0],fsplit[1],fsplit[4][:-4]])+"_z_linkage_array.npy", z)
        
```

```python
def compare(sess1, sess2):
    clist = []
    for i, g1 in enumerate(sess1):
        for j, g2 in enumerate(sess2):
            tmp_top = set(g1.values) & set(g2.values)
            tmp_bottom = set(g1.values) ^ set(g2.values)
            c = len(tmp_top)/len(tmp_bottom) if len(tmp_bottom)>0 else 0
            clist.append([i,j,c])
    return clist
```
