```python
walker = os.walk("C:\\Users\\Mango\\Documents\\Complexity\\Midterm Project\\output")
```

```python
txt_f = []
for curr, dirs, files in walker:
    for f in files:
        if f[-3:]=='bak':
            txt_f.append(os.path.join(curr,f))
```

```python
mod_files = []
for f in txt_f:
    if os.path.getsize(f)>0:
         mod_files.append(f)
```

```python
for f in mod_files:
    with open(f, 'r') as rf, open(f[:-4], 'w') as wf:
        rl = rf.readline()
        while rl!='':
            homogenized_l = rl.replace("\\", "/")
            if rl[-2:]=='\n':
                print("Flagging newline!\nCheck fidelity!")
                break
            wf.write(homogenized_l)
            rl = rf.readline()
```
